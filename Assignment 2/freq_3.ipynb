```python
def create_kernel(im):
    """
    Creates a filter kernel in the frequency domain.

    Args:
        im: np.array of shape [H, W]
    Returns:
        kernel: np.array of shape [H, W]
    """
    ### START YOUR CODE HERE ### (You can change anything inside this block)
    low_pass_filter = utils.create_low_pass_frequency_kernel(im, radius=29)
    high_pass_filter = utils.create_high_pass_frequency_kernel(im, radius=30)
    kernel = low_pass_filter + high_pass_filter
    ### END YOUR CODE HERE ###
    return kernel
```
